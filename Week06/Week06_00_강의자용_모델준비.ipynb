```python
# 1. Google Drive 연결 및 저장 폴더 확인
import os
from google.colab import drive
drive.mount("/content/drive")

TARGET = "/content/drive/MyDrive/2026-2강의/동서울대학교/__인공지능융합활용_수789/6주_강의자료/AI융합활용_6주_모델/ko-sroberta-multitask"
MODEL_FOLDER_ID = "1EmofMTJ6HRQrth8puRpHwvFDJ27YeiVr"

if os.path.isdir(TARGET):
    print("저장 폴더 확인:", TARGET)
    print("현재 들어 있는 파일:", os.listdir(TARGET) or "(비어 있음)")
else:
    raise FileNotFoundError("저장 폴더가 없습니다. 경로를 확인하세요. (폴더를 새로 만들면 ID가 바뀌어 학생 노트북 수정이 필요합니다)")
```

```python
# 2. 모델 내려받기 → Drive 폴더에 저장 (Colab 서버에서 실행되므로 강의실 네트워크와 무관)
import importlib.util, shutil, time
if importlib.util.find_spec("sentence_transformers") is None:
    !pip install -q sentence-transformers
from sentence_transformers import SentenceTransformer

start = time.time()
model = SentenceTransformer("jhgan/ko-sroberta-multitask")
model.save("/content/ko-sroberta-multitask")          # 실행에 필요한 파일만 저장
shutil.copytree("/content/ko-sroberta-multitask", TARGET, dirs_exist_ok=True)
print(f"저장 완료 ({time.time() - start:.0f}초)")

total = 0
for root, _, files in os.walk(TARGET):
    for f in files:
        size = os.path.getsize(os.path.join(root, f))
        total += size
        print(f"  {os.path.relpath(os.path.join(root, f), TARGET):40s} {size / 1e6:8.1f} MB")
print(f"합계: {total / 1e6:.0f} MB")
```

```python
# 3. Drive에 저장된 모델로 실제로 불러지는지 확인 후, 업로드 마무리
test_model = SentenceTransformer(TARGET)
v = test_model.encode(["등록금 납부 기간이 언제인가요?", "학비는 언제까지 내야 하나요?"])
sim = float(v[0] @ v[1] / ((v[0] @ v[0]) ** 0.5 * (v[1] @ v[1]) ** 0.5))
print("벡터 차원:", v.shape[1], "/ 예시 유사도:", round(sim, 3))

drive.flush_and_unmount()     # Drive 업로드가 끝날 때까지 기다림
print("Drive 동기화 완료")
```

```python
# 4. 학생과 같은 방법(공유 링크 복사)으로 내려받아지는지 시험
import gdown
shutil.rmtree("/content/student_test", ignore_errors=True)
start = time.time()
gdown.download_folder(id=MODEL_FOLDER_ID, output="/content/student_test", quiet=True)
print(f"학생용 링크 복사 시간: {time.time() - start:.0f}초")
SentenceTransformer("/content/student_test")
print("✅ 학생 노트북에서 사용할 준비가 끝났습니다.")
```
