## 목표
- Teachable Machine에서 Export한 학습 결과 모델을 Colab으로 가져온다.
- 모델 입력 크기와 클래스 이름을 확인한다.
- 새로운 이미지에 대해 추론하고 예측 확률을 분석한다.
- 오분류 원인을 기록한다.


```python
from google.colab import files
uploaded = files.upload()
print(uploaded.keys())
```

```python
import tensorflow as tf
MODEL_FILE = "keras_model.h5"
LABEL_FILE = "labels.txt"
model = tf.keras.models.load_model(MODEL_FILE, compile=False)
with open(LABEL_FILE, "r", encoding="utf-8") as f:
    class_names = [line.strip() for line in f if line.strip()]
print("Classes:", class_names)
print("Input shape:", model.input_shape)
```

```python
test_uploaded = files.upload()
print(test_uploaded.keys())
```

```python
import numpy as np
from PIL import Image
IMAGE_FILE = next(iter(test_uploaded.keys()))
height, width = model.input_shape[1], model.input_shape[2]
img = Image.open(IMAGE_FILE).convert("RGB").resize((width, height))
x = np.asarray(img).astype(np.float32)
x = (x / 127.5) - 1
x = np.expand_dims(x, axis=0)
pred = model.predict(x, verbose=0)[0]
best = int(np.argmax(pred))
print("Prediction:", class_names[best])
print("Confidence:", float(pred[best]))
for name, score in zip(class_names, pred):
    print(f"{name}: {score:.4f}")
```
