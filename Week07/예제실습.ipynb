
# 리스트 (List)

- 순서가 있고 수정이 가능한 자료형입니다.
- 중복된 값을 허용합니다.

```python
# 리스트 생성 및 요소를 추가/수정
fruits = ["apple", "banana", "cherry"] # 0 1 2
fruits.append("orange")  # 요소 추가 # 0 1 2 3
fruits[1] = "blueberry"  # 요소 수정

print(fruits)  # 출력: ['apple', 'blueberry', 'cherry', 'orange']
```


# 딕셔너리 (Dictionary)

- 키(Key)와 값(Value)의 쌍으로 이루어진 자료형입니다.
- 키는 중복될 수 없습니다.

```python
# 딕셔너리 생성 및 데이터 접근/추가
user = {"name": "Alice", "age": 25}
user["age"] = 26          # 값 수정
user["city"] = "Seoul"    # 새로운 키-값 쌍 추가

print(user)       # 출력: {'name': 'Alice', 'age': 26, 'city': 'Seoul'}
print(user["name"]) # 출력: Alice
```


# 튜플 (Tuple)

- 리스트와 유사하지만,
- 한 번 생성하면 요소를 변경하거나 삭제할 수 없습니다 (변경 불가능).

```python
# 튜플 생성 및 접근 (수정 시도 시 에러 발생)
coordinates = (10, 20, 30) # 0 1 2

print(coordinates[0]) # 출력: 10

# coordinates[0] = 40  # TypeError 발생 (수정 불가)
```


# 🎯 세트 (Set)

- 순서가 없고
- 중복을 허용하지 않는 자료형입니다
- 집합 연산에 유용합니다.

```python
# 세트 생성 및 중복 제거 확인
numbers = {1, 2, 2, 3, 3, 3} # {1, 2, 3}
numbers.add(4)  # 요소 추가   {1, 2, 3, 4}

print(numbers)  # 출력: {1, 2, 3, 4} (중복 자동 제거)
```

# def : function만들기
-  단어를 3개씩 읽어서 출력하는 함수 만들기


```python
def split_words(text, size=3):
    words = text.split()
    return [
        " ".join(words[i:i + size])
        for i in range(0, len(words), size)
    ]

words_text = "PDF 문서를 읽고 텍스트를 검색합니다"
print(words_text)
print(split_words("PDF 문서를 읽고 텍스트를 검색합니다", 3))
```

# 문장을 만들고 cosine 일치도를 출력하기




```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "학교의 안전관리 계획과 절차",
    "Gradio를 이용한 웹 챗봇",
    "장애아동 지원체계 조사 결과"
]
vectorizer = TfidfVectorizer(analyzer="char", ngram_range=(2, 4))
matrix = vectorizer.fit_transform(documents)
query = vectorizer.transform(["안전관리 절차"])
scores = cosine_similarity(query, matrix)[0]

for idx, score in enumerate(scores):
    print(idx, round(float(score), 3), documents[idx])
```
